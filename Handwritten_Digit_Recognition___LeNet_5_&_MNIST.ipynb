```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.layers import Conv2D, AveragePooling2D, Dense, Flatten
from tensorflow.keras.models import Sequential
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
import matplotlib.pyplot as plt
(x_train , y_train), (x_test, y_test) = mnist.load_data()
x_train.shape
x_train = np.pad(x_train,((0,0),(2,2),(2,2)))
x_test  = np.pad(x_test,((0,0),(2,2),(2,2)))
x_train = x_train/255.0
x_test  = x_test/255.0
y_train = to_categorical(y_train,10)
y_test  = to_categorical(y_test,10)
#after add padding
x_train.shape
plt.imshow(x_train[5], cmap='gray')
plt.show()
x_train = np.expand_dims(x_train, 3)
x_test  = np.expand_dims(x_test, 3)
#LeNet expects images of size 32x32x1 instead of 32x32
lenet = Sequential(name="LeNet-5")

lenet.add(Conv2D(6,(5,5),strides=(1,1), activation='tanh', input_shape=(32,32,1), name='C1')) # C1
lenet.add(AveragePooling2D(name='S2', pool_size=(2,2))) # S2

lenet.add(Conv2D(16,(5,5),strides=(1,1), activation='tanh', name='C3')) # C3
lenet.add(AveragePooling2D(name='S4', pool_size=(2,2))) # S4

lenet.add(Flatten())

lenet.add(Dense(120,activation='tanh', name='FC5'))      # FC5
lenet.add(Dense(84,activation='tanh', name='FC6'))       # FC6

lenet.add(Dense(10,activation='softmax', name='Output')) # FC7
lenet.summary()
!pip install visualkeras
import visualkeras
visualkeras.layered_view(lenet, legend=True)
lenet.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
# categorical_crossentropy for multi-class , binary_crossentropy for binary class
history=lenet.fit(x_train, y_train, epochs=2, batch_size=32)
_, acc = lenet.evaluate(x_test, y_test)
plt.title('Training Log')
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Score')
plt.legend()
```
